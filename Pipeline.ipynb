Importing libraries and Datasets

```python
import numpy as np
import pandas as pd
```

```python
#Uploading the Dataset
df = pd.read_csv('Titanic-Dataset.csv')
```

```python
#Renaming the colomns name for Modern Data
df = df.rename(columns={'Age':'Age','Fare':'Cost_USD','Pclass':'Service_Tier'})
```

```python
#(rows,Columns)
df.shape
```

```python
#display the first 10 rows
df.head(10)
```

```python
#missing values in the Each columns
df.isna().sum()
```

```python
#Calculating the threshold value
total_rows = 891
missing_threshold = total_rows * 0.70
print(f'Maximum missing values allowed: {missing_threshold}')
```

```python
#dropping the colomn Cabin it has too many missing values
df.drop(columns=['Cabin'], inplace=True)
```

```python
#Finding the medain value of Age
median_Age = np.nanmedian(df['Age'])
print(f'Median Age : {median_Age}')
```

```python
#Filling the nan values 
df['Age'] = df['Age'].fillna(median_Age)
df
```

```python
#Giving the value with the Unknown
df['Embarked'] = df['Embarked'].fillna('Unknown')
```

```python
#Exact rows with NaN in Embarked has been changed to Unknown
df.loc[[61,829]]
```

```python
#Checking Again to Conform No Null values
df.isna().sum()
```

```python
#New Column Created for Cost in INR
df['Cost_INR'] = df['Cost_USD'] * 83.5
df
```

```python


# Corrected parenthetical nesting
df['Age_Group'] = np.where(
    df['Age'] < 18, 
    'Minor', 
    np.where((df['Age'] >= 18) & (df['Age'] < 60), 'Adult', 'Senior')
)
df
```

```python
#Age Category
df['Age_Group'].value_counts()
```

```python
#Cost_usd and Age ,sum and mean value
df.groupby('Service_Tier')[['Cost_USD','Age']].agg(['sum','mean'])
```

```python
#Amount spend by each age group
df.groupby('Age_Group')['Cost_USD'].max()
```
